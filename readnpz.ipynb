```python
import numpy as np
import pandas as pd

def get_similar_movies_with_scores(movie_index, top_k=5):
    """
    Return the top_k most similar movie titles to a given movie index,
    along with similarity scores and directors.

    Parameters:
        movie_index (int): Index of the reference movie.
        top_k (int): Number of similar results to return.

    Returns:
        DataFrame with 'title', 'director', and 'score' columns.
    """

    # Load saved FAISS indices and scores
    indices = np.load('director/indices.npz')['arr_0']
    scores = np.load('director/scores.npz')['arr_0']

    # Load the movies metadata
    movies = pd.read_csv('final.csv')

    # Get top-k similar indices and their scores
    similar_indices = indices[movie_index][:top_k]
    similar_scores = scores[movie_index][:top_k]

    # Gather movie info
    similar_movies = movies.iloc[similar_indices].copy()
    similar_movies['score'] = similar_scores

    # Select only the relevant columns
    return similar_movies[['title', 'director', 'score']]
```

```python
similar = get_similar_movies_with_scores(movie_index=0, top_k=5)

print("Top similar movies:")
print(similar)
```
